```python
ruta_2024 = "/Volumes/workspace/pobreza/raw/ECH_2024_final.csv"
ruta_2025 = "/Volumes/workspace/pobreza/raw/ECH_2025_implantacion.csv"

def leer(ruta):
    return (spark.read
            .option("header", True)
            .option("inferSchema", True)
            .option("sep", ",")
            .option("encoding", "UTF-8")
            .csv(ruta))

ech24 = leer(ruta_2024)
ech25 = leer(ruta_2025)

print("2024:", ech24.count(), "filas,", len(ech24.columns), "columnas")
print("2025:", ech25.count(), "filas,", len(ech25.columns), "columnas")
print(ech24.columns[:30])
display(ech24.limit(5))
```

```python
for df, nombre in [(ech24, "ech_2024_bronze"), (ech25, "ech_2025_bronze")]:
    (df.drop("_c0")
       .write.mode("overwrite")
       .option("overwriteSchema", True)
       .saveAsTable(f"workspace.pobreza.{nombre}"))
    print("Guardada:", nombre)
```
